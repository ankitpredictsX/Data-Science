# **Stacking**

```python
a4= np.arange(12).reshape(3,4)
a5= np.arange(12,24).reshape(3,4)
print(1,np.hstack((a4,a5)))#along the existing axis
print(2,np.vstack((a4,a5)))
print(3,np.stack((a4,a5),axis=1))#default axis=0
print(4,np.stack((a4,a5)))
print(5,np.dstack((a4,a5)))#dstack stacks arrays along the third axis (depth axis)
#np.dstack((a, b)).shape
# (m, n, 2)
result = np.column_stack((a4, a5))#same as np.hstack() for 2d
print(6,result)
```

# Set **Operations**

```python
#union1d
a = np.array([1,2,3,4,5])
b = np.array([3,4,5,6,7])
np.union1d(a,b)
```

```python
#intersection1d
a = np.array([1,2,3,4,5])
b = np.array([3,4,5,6,7])
np.intersect1d(a,b)
```

```python
#setdiff1d->Elements in first array but not in second
a = np.array([1,2,3,4,5])
b = np.array([3,4,5,6,7])
np.setdiff1d(a,b)
```

```python
#setxor1d()->Symmetric difference = elements in either array, but not both
a = np.array([1,2,3,4,5])
b = np.array([3,4,5,6,7])
np.setxor1d(a,b)
```

```python
# fill the nan values of array with most common element in array
arr = np.array([[1,2,np.nan],
                [4,2,6],
                [np.nan,np.nan,5]])
d = {}
for i in arr.flatten():
    if not np.isnan(i):      # Skip NaN values
        if i in d:
            d[i] += 1
        else:
            d[i] = 1

most_common = sorted(d.items(), key=lambda x: x[1], reverse=True)[0][0]

arr[np.isnan(arr)] = most_common

print(arr)
```
