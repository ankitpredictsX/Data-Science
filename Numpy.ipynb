```python
# fill the nan values of array with most common element in array
arr = np.array([[1,2,np.nan],
                [4,2,6],
                [np.nan,np.nan,5]])
d = {}
for i in arr.flatten():
    if not np.isnan(i):      # Skip NaN values
        if i in d:
            d[i] += 1
        else:
            d[i] = 1

most_common = sorted(d.items(), key=lambda x: x[1], reverse=True)[0][0]

arr[np.isnan(arr)] = most_common

print(arr)
#or
arr = np.array([[1,2,np.nan],
                [4,2,6],
                [np.nan,np.nan,5]])

# Remove NaNs
valid = arr[~np.isnan(arr)]

# Find unique values and counts
values, counts = np.unique(valid, return_counts=True)

# Most common value
most_common = values[np.argmax(counts)]

# Fill NaNs
arr[np.isnan(arr)] = most_common

print(arr)
```

```python
#Wap implementing broadcasting rule to check 2 array can be added or not taking 2 tuples as a input dimension
import numpy as np

def can_broadcast(s1, s2):
    r1 = s1[::-1]
    r2 = s2[::-1]

    # Zip stops at the shorter — that's correct per broadcasting rules
    result = []
    for i, j in zip(r1, r2):
        if i != j and i != 1 and j != 1:
            return False, None
        result.append(max(i, j))

    # Append remaining dims from the longer shape
    result.extend(r1[len(r2):])
    result.extend(r2[len(r1):])

    return True, tuple(result[::-1])


# ── Test ────────────────────────────────────────────────────
test_cases = [
    ((3, 4),    (3, 4)),
    ((5, 1),    (5,)),
    ((3, 1, 4), (3, 5, 4)),
    ((2, 3),    (4, 3)),
    ((1,),      (9, 9, 9)),
]

for s1, s2 in test_cases:
    ok, out_shape = can_broadcast(s1, s2)
    if ok:
        np_shape = np.broadcast_shapes(s1, s2)
        match = "✓" if out_shape == np_shape else "✗ mismatch"
        print(f"{str(s1):15} + {str(s2):15} → {str(out_shape):15} {match}")
    else:
        print(f"{str(s1):15} + {str(s2):15} → Cannot broadcast   ✗")
```

```python
a = np.random.randint(0, 50, (3, 4))
b = np.random.randint(0, 50, (3, 4))

print("Array a:"); print(a)
print("Array b:"); print(b)

# ── Method 1: np.maximum (most direct) ─────────────────────
print("\nMethod 1 — np.maximum:")
print(np.maximum(a, b))

# ── Method 2: np.where ─────────────────────────────────────
print("\nMethod 2 — np.where:")
print(np.where(a > b, a, b))

# ── Method 3: stacking + max ───────────────────────────────
print("\nMethod 3 — np.stack + max:")
print(np.stack([a, b]).max(axis=0))

# ── Method 4: boolean masking ──────────────────────────────
print("\nMethod 4 — boolean masking:")
result = a.copy()
result[b > a] = b[b > a]
print(result)
```
